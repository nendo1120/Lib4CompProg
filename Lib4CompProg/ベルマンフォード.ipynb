```python
# repository: https://github.com/nendo1120/Lib4CompProg

#ベルマンフォード法
def BellmanFord(edges,num_v,s):
    #グラフの初期化
    inf=10**19
    dist=[inf]*num_v
    dist[s]=0
  
    #辺の緩和
    for i in range(num_v):
        for edge in edges: #edge=[頂点,頂点,重み]
            if edge[0] != inf and dist[edge[1]] > dist[edge[0]] + edge[2]:
                dist[edge[1]] = dist[edge[0]] + edge[2]
                if i==num_v-1 and edge[1]==num_v-1: return -1
  
    return dist
```
