```python
# repository: https://github.com/nendo1120/Lib4CompProg

from random import randint
mod = (1<<61)-1
base = randint(10007,mod-1)
class RollingHash():
    def __init__(self, s, base, mod):
        self.mod = mod
        self.pw = pw = [1]*(len(s)+1)
        self.base = base

        l = len(s)
        self.h = h = [0]*(l+1)

        v = 0
        for i in range(l):
            h[i+1] = v = (v * self.base + s[i]) % mod
        v = 1
        for i in range(l):
            pw[i+1] = v = v * self.base % mod
    def get(self, l, r): # 1-indexed # (l,r]
        return (self.h[r] - self.h[l] * self.pw[r-l]) % self.mod
```
