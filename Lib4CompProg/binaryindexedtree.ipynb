```python
# Binary Indexed Tree
# 1-indexed
class BinaryIndexedTree():
    def __init__(self,n):
        self.size=n
        self.tree=[0]*(n+1)
        self.depth=n.bit_length()
        
    def query(self,idx):
        s=0
        while idx>0:
            s+=self.tree[idx]
            idx-=idx&(-idx)
        return s
    
    def add(self,idx,x):
        while idx<=self.size:
            self.tree[idx]+=x
            idx+=idx&(-idx)
    
    def search(self,x): # a_1+a_2+...+a_idx>=xとなる最小のidxを求める。
        s,idx=0,0
        for i in reversed(range(self.depth)):
            k=idx+(1<<i)
            if k<=self.size and s+self.tree[k]<x:
                s+=self.tree[k]
                idx+=1<<i
        return idx+1
```
