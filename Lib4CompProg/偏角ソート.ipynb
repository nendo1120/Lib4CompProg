```python
# https://github.com/nendo1120/Lib4CompProg/blob/main/Lib4CompProg/%E5%81%8F%E8%A7%92%E3%82%BD%E3%83%BC%E3%83%88.ipynb

# name: Cmp to Key
# prefix: cmp to key
# ---
# 拝借
# https://github.com/manuo-git/python-lib/blob/main/src/cmp_to_key.py
from typing import *

def sort(data: list[Any], cmp):
    if len(data) == 1: return
    res = _sort(data, cmp)
    data.clear()
    data.extend(res)

def _sort(data: list[Any], cmp):
    if len(data) <= 1:
        return data
    m = len(data)//2
    ret = merge(_sort(data[:m], cmp), _sort(data[m:], cmp), cmp)
    return ret

def merge(l, r, cmp):
    res = []
    i, j = 0, 0
    while i < len(l) and j < len(r):
        if cmp(l[i], r[j]):
            res.append(l[i])
            i += 1
        else:
            res.append(r[j])
            j += 1
    if i < len(l): res.extend(l[i:])
    if j < len(r): res.extend(r[j:])
    return res

# 拝借
# https://atcoder.jp/contests/abc481/editorial/27922
def cmp(a, b):
    # aが左ならTrue
    x, y = a
    u, v = b
    h = y < 0 or (y == 0 and x < 0)
    k = v < 0 or (v == 0 and u < 0)
    if h != k:
        return not h
    z = x * v - y * u
    return z > 0

points=[]
points.sort(key=cmp_to_key(cmp))
```
