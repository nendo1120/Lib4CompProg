```python
# 拝借
# https://atcoder.jp/contests/abc481/editorial/27922

from functools import cmp_to_key

def cmp(a, b):
        x, y = a
        u, v = b
        h = y < 0 or (y == 0 and x < 0)
        k = v < 0 or (v == 0 and u < 0)
        if h != k:
            return h - k
        z = x * v - y * u
        return (z < 0) - (z > 0)
points=[]
points.sort(key=cmp_to_key(cmp))
```
