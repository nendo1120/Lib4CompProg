```python
from bisect import bisect_left

# lis[i]:= 長さiの増加部分列の末尾の値の最小値
def calc_LIS(arr):
    n=len(arr)
    lis=[arr[0]]

    for i in range(n):
        if lis[-1]<arr[i]:
            lis.append(arr[i])
        else:
            idx=bisect_left(lis,arr[i])
            lis[idx]=arr[i]
    
    return len(lis)
```
