```python
diagonal=1
non_diagonal=0
# 単位行列の成分
# (×, +)：(1, 0)
# (+, min)：(0, inf)
# (+, max)：(0, -inf)

#a*bの計算
#n=lの必要あり
def mat_mul(a,b,non_diagonal,mod):
    n,m,l=len(a),len(b[0]),len(b)
    c=[[non_diagonal]*m for _ in range(n)]
    for i in range(n):
        for k in range(l):
            for j in range(m):
                c[i][j]=(c[i][j]+a[i][k]*b[k][j])%mod
    return c

#a^nの計算
def mat_pow(a,n,diagonal,non_diagonal,mod):
    m=len(a)
    b=[[non_diagonal]*m for _ in range(m)]
    for i in range(m):
        b[i][i]=diagonal
    while n>0:
        if n&1:
            b=mat_mul(b,a,mod)
        a=mat_mul(a,a,mod)
        n>>=1
    return b
```
